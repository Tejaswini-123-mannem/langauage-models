```python
import re

def script_stats(text):
    telugu = len(re.findall(r'[\u0C00-\u0C7F]', text))
    latin = len(re.findall(r'[A-Za-z]', text))
    digits = len(re.findall(r'\d', text))
    total = len(text)

    return telugu, latin, digits, total


sample = dataset["train"][:10000]["text"]

telugu_chars = 0
latin_chars = 0
digit_chars = 0
total_chars = 0

for text in sample:
    t, l, d, total = script_stats(text)
    telugu_chars += t
    latin_chars += l
    digit_chars += d
    total_chars += total

print("Total characters:", total_chars)
print("Telugu characters:", telugu_chars)
print("Latin characters:", latin_chars)
print("Digits:", digit_chars)

print("Telugu ratio:", telugu_chars / total_chars)
print("Latin ratio:", latin_chars / total_chars)
print("Digit ratio:", digit_chars / total_chars)
```

```python
from collections import Counter
import unicodedata

sample = dataset["train"][:10000]["text"]

script_counts = Counter()

for text in sample:
    for ch in text:
        if ch.isspace():
            category = "whitespace"
        elif ch.isdigit():
            category = "digit"
        elif '\u0C00' <= ch <= '\u0C7F':
            category = "Telugu"
        elif ('A' <= ch <= 'Z') or ('a' <= ch <= 'z'):
            category = "Latin"
        elif unicodedata.category(ch).startswith("P"):
            category = "punctuation"
        else:
            category = "other"

        script_counts[category] += 1

total = sum(script_counts.values())

for category, count in script_counts.most_common():
    print(f"{category:15} {count:10} {count/total:.4%}")
```

```python
import re
import unicodedata

def clean_text(text):
    # 1. Unicode normalization
    text = unicodedata.normalize("NFC", text)

    # 2. Remove speech/video timestamp metadata
    text = re.sub(
        r'\(Refer Slide Time:\s*\d+:\d+\)',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # 3. Remove "Slide N" metadata and an optional following |
    text = re.sub(
        r'\bSlide\s+\d+\s*\|?',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # 4. Normalize spaces and tabs
    text = re.sub(r'[ \t]+', ' ', text)

    # 5. Normalize excessive blank lines
    text = re.sub(r'\n{3,}', '\n\n', text)

    # 6. Remove spaces immediately before common punctuation
    text = re.sub(r'\s+([,.!?;:])', r'\1', text)

    # 7. Strip
    text = text.strip()

    return text
```

```python
from collections import Counter
import re

def repeated_paragraph_stats(text):
    # Split roughly on paragraph/newline boundaries
    paragraphs = [
        p.strip()
        for p in re.split(r'\n+', text)
        if p.strip()
    ]

    if len(paragraphs) <= 1:
        return 0, 0

    counts = Counter(paragraphs)

    repeated = sum(
        count - 1
        for count in counts.values()
        if count > 1
    )

    return repeated, len(paragraphs)


total_paragraphs = 0
repeated_paragraphs = 0
docs_with_repetition = 0
docs_checked = 0

for i in range(10000):
    row = dataset["train"][i]

    if row["type"] != "web":
        continue

    repeated, total = repeated_paragraph_stats(row["text"])

    total_paragraphs += total
    repeated_paragraphs += repeated

    if repeated > 0:
        docs_with_repetition += 1

    docs_checked += 1

    if docs_checked == 1000:
        break

print("Web documents checked:", docs_checked)
print("Total paragraphs:", total_paragraphs)
print("Repeated paragraphs:", repeated_paragraphs)
print("Documents containing repetition:", docs_with_repetition)

if total_paragraphs:
    print(
        "Repeated paragraph rate:",
        repeated_paragraphs / total_paragraphs * 100,
        "%"
    )

print(
    "Document repetition rate:",
    docs_with_repetition / docs_checked * 100,
    "%"
)
```

```python
import re

def telugu_ratio(text):
    total_letters = 0
    telugu_letters = 0

    for ch in text:
        if ch.isalpha():
            total_letters += 1

            if '\u0C00' <= ch <= '\u0C7F':
                telugu_letters += 1

    if total_letters == 0:
        return 0.0

    return telugu_letters / total_letters


sample = dataset["train"][:10000]["text"]

ratios = [telugu_ratio(text) for text in sample]

print("Minimum Telugu ratio:", min(ratios))
print("Maximum Telugu ratio:", max(ratios))
print("Average Telugu ratio:", sum(ratios) / len(ratios))
```

```python
def is_valid_telugu(text):
    if not text:
        return False

    # Avoid extremely short records
    if len(text.strip()) < 50:
        return False

    # Preliminary Telugu-language filter
    if telugu_ratio(text) < 0.10:
        return False

    return True
```

```python
import hashlib
import json
import os
import gc

DEDUP_DIR = (
    "/content/drive/MyDrive/"
    "LMA_Individual_Project/data/processed/sangraha"
)

dedup_output = f"{DEDUP_DIR}/sangraha_deduplicated.jsonl"
hash_checkpoint = f"{DEDUP_DIR}/dedup_hashes.txt"

seen_hashes = set()

total_records = 0
unique_records = 0
duplicate_records = 0

# Process one shard at a time
with open(dedup_output, "w", encoding="utf-8") as out:

    for shard_idx in range(4):

        input_path = (
            f"{DEDUP_DIR}/"
            f"sangraha_shard_{shard_idx:02d}.jsonl"
        )

        print("\n" + "=" * 70)
        print(f"DEDUPLICATING SHARD {shard_idx}")
        print("=" * 70)

        shard_records = 0
        shard_duplicates = 0

        with open(input_path, "r", encoding="utf-8") as f:

            for line in f:

                row = json.loads(line)

                text = row["text"]

                # Exact hash of already-cleaned text
                text_hash = hashlib.sha256(
                    text.encode("utf-8")
                ).hexdigest()

                total_records += 1
                shard_records += 1

                if text_hash in seen_hashes:
                    duplicate_records += 1
                    shard_duplicates += 1
                    continue

                seen_hashes.add(text_hash)
                unique_records += 1

                out.write(
                    json.dumps(
                        row,
                        ensure_ascii=False
                    ) + "\n"
                )

        print(f"Records processed: {shard_records:,}")
        print(f"Duplicates found:  {shard_duplicates:,}")
        print(f"Unique so far:     {unique_records:,}")

        # Force write to disk
        out.flush()

        # Save hash checkpoint
        with open(hash_checkpoint, "w", encoding="utf-8") as hf:
            for h in seen_hashes:
                hf.write(h + "\n")

        print("Hash checkpoint saved.")

        gc.collect()

print("\n" + "=" * 70)
print("GLOBAL EXACT DEDUPLICATION COMPLETE")
print("=" * 70)

print(f"Total records:      {total_records:,}")
print(f"Unique records:     {unique_records:,}")
print(f"Exact duplicates:   {duplicate_records:,}")

print(
    f"Duplicate rate:     "
    f"{duplicate_records / total_records * 100:.4f}%"
)

print(
    f"Retention rate:     "
    f"{unique_records / total_records * 100:.4f}%"
)

print(f"\nOutput: {dedup_output}")
print(f"Hashes: {hash_checkpoint}")
```

```python
import json
import random

DEDUP_FILE = (
    "/content/drive/MyDrive/"
    "LMA_Individual_Project/data/processed/"
    "sangraha/sangraha_deduplicated.jsonl"
)

random.seed(42)

sample_records = []

with open(DEDUP_FILE, "r", encoding="utf-8") as f:
    for line in f:
        if len(sample_records) >= 10000:
            break

        sample_records.append(json.loads(line))

print("Sample records:", len(sample_records))
```

```python
def get_shingles(text, k=5):
    words = text.split()

    if len(words) < k:
        return set(words)

    return {
        " ".join(words[i:i+k])
        for i in range(len(words) - k + 1)
    }
```

```python
from datasketch import MinHash, MinHashLSH

NUM_PERM = 128
THRESHOLD = 0.90

lsh = MinHashLSH(
    threshold=THRESHOLD,
    num_perm=NUM_PERM
)
```

```python
minhashes = {}
candidate_pairs = []

for idx, row in enumerate(sample_records):

    text = row["text"]

    # Don't near-deduplicate very short documents
    if len(text) < 200:
        continue

    shingles = get_shingles(text, k=5)

    if not shingles:
        continue

    m = MinHash(num_perm=NUM_PERM)

    for shingle in shingles:
        m.update(shingle.encode("utf-8"))

    # Query BEFORE inserting this document
    candidates = lsh.query(m)

    if candidates:
        candidate_pairs.append(
            (idx, candidates)
        )

    lsh.insert(str(idx), m)
    minhashes[idx] = m

print("Documents with near-duplicate candidates:",
      len(candidate_pairs))
```

```python
for idx, candidates in candidate_pairs[:20]:

    print("\n" + "=" * 80)

    print("DOCUMENT A:", idx)
    print(sample_records[idx]["text"][:500])

    for candidate_id in candidates[:3]:

        candidate_id = int(candidate_id)

        print("\nDOCUMENT B:", candidate_id)
        print(sample_records[candidate_id]["text"][:500])
```

```python
THRESHOLD = 0.80

lsh_80 = MinHashLSH(
    threshold=THRESHOLD,
    num_perm=128
)

candidate_pairs_80 = []
minhashes_80 = {}

for idx, row in enumerate(sample_records):

    text = row["text"]

    if len(text) < 200:
        continue

    shingles = get_shingles(text, k=5)

    if not shingles:
        continue

    m = MinHash(num_perm=128)

    for shingle in shingles:
        m.update(shingle.encode("utf-8"))

    candidates = lsh_80.query(m)

    if candidates:
        candidate_pairs_80.append((idx, candidates))

    lsh_80.insert(str(idx), m)
    minhashes_80[idx] = m

print(
    "Candidates at 80%:",
    len(candidate_pairs_80)
)
```
